We combine Last.fm listening history with Spotify audio features to study user listening diversity.

## Research Question

**How do users’ listening habits relate to their listening genre diversity?**

### Why this matters:
- Users vary widely in how many different **genres** they explore.
- Spotify provides rich audio features that describe musical characteristics.
- Combining Last.fm listening data with Spotify metadata reveals deeper insights into user taste.
- Understanding genre diversity can help improve personalization and recommendation systems.



## Data Collection & Cleaning

- **Last.fm data:** U.S. users with 150,000+ scrobbles (high-activity listeners).

## Measuring Listening Diversity

We define **genre diversity** as the number of unique genres a user listens to across their top tracks.

Most users listen to between 5 to 10 genres.




## Spotify Feature Relationships

Spotify audio features show meaningful structure:

- **Energy** & **Danceability** are positively correlated.
- **Acousticness** is negatively correlated with Energy.
- Similarity scores cluster together.

These insights help explain user listening patterns.




## High vs Low Genre Diversity

We split users at the median genre diversity.

Key observations:
- High-diversity users tend to listen to slightly more danceable and energetic tracks.
- Low-diversity users show narrower listening patterns.
